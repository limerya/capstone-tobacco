```python
import rasterio, numpy as np, glob, os

F = r"D:\Projects\capstone"
find = lambda n: glob.glob(os.path.join(F, "**", n), recursive=True)[0]

def land_stats(fname, bands, land_rule):
    with rasterio.open(find(fname)) as s:
        a = s.read().astype(float)
        names = list(s.descriptions)
    land = land_rule(a, names)
    print(f"\n{fname}  (land pixels: {land.sum()})")
    for b in bands:
        v = a[names.index(b)][land]
        print(f"  {b:15s} p5={np.percentile(v, 5):9.1f}  p50={np.median(v):9.1f}  "
              f"p95={np.percentile(v, 95):9.1f}  min={v.min():9.1f}  max={v.max():9.1f}")

# climate: outside-AOI pixels are zero in every band
land_stats("climate_REGION_1_2021_2022.tif", ["precip_cv", "soil_moist"],
           lambda a, n: a.any(axis=0))

# soil: real soil never has sand + clay + silt = 0
land_stats("soil_REGION_1.tif", ["drainage_proxy"],
           lambda a, n: (a[n.index("sand")] + a[n.index("clay")] + a[n.index("silt")]) > 0)
```

```python
import pandas as pd, numpy as np
df = pd.read_parquet(r"D:\Projects\capstone\training_sample_labeled.parquet")
d = df[df["fao_class"].notna()]

doms = [c for c in d.columns if c.startswith("dom_")]
print("mean domain rating:")
print(d[doms].mean().round(1).sort_values().to_string())

print("\nwhich domain is limiting (the minimum), all scored cells:")
print(d[doms].idxmin(axis=1).value_counts(normalize=True).mul(100).round(1).to_string())

# counterfactual: pH optimum widened to the 5.5-6.5 your Table 3.2 cites
WIDE = [(3.5,5),(5.0,25),(5.2,45),(5.4,70),(5.5,98),(6.5,98),(6.8,70),(7.2,45),(7.8,25),(9.0,5)]
xs = np.array([p[0] for p in WIDE]); ys = np.array([p[1] for p in WIDE])
ph_new = np.interp(d["ph"], xs, ys)

fert = np.fmin.reduce(np.vstack([ph_new, d["rate_soc"], d["rate_cec"]]))
D = np.vstack([d["dom_climate"], d["dom_topography"], d["dom_wetness"],
               d["dom_physical_soil"], fert])
rmin = np.nanmin(D, axis=0)
order = np.argsort(np.where(np.isfinite(D), D, np.inf), axis=0)
others = np.take_along_axis(D, order[1:], axis=0) / 100.0
idx = rmin * np.sqrt(np.where(np.isfinite(others), others, 1.0).prod(axis=0))

for lab, v in [("current", d["index_sqrt"]), ("wider pH", idx)]:
    v = np.asarray(v, float)
    print(f"\n{lab:9s} S1 {100*(v>=75).mean():.3f}%  S2 {100*((v>=50)&(v<75)).mean():.2f}%"
          f"  S3 {100*((v>=25)&(v<50)).mean():.2f}%  N {100*(v<25).mean():.2f}%"
          f"  | p99={np.nanpercentile(v,99):.1f} max={np.nanmax(v):.1f}")
```

```python
import pandas as pd, numpy as np
d = pd.read_parquet(r"D:\Projects\capstone\training_sample_labeled.parquet")
d = d[d["fao_class"].notna()]

F = {"climate": ["temp","sunshine"], "topography": ["slope"], "wetness": ["drainage"],
     "physical_soil": ["texture"], "fertility": ["ph","soc","cec"]}
R = {k: np.vstack([d[f"rate_{f}"].to_numpy(float) for f in v]) for k, v in F.items()}

def gmean(a): return np.exp(np.nanmean(np.log(np.clip(a, 1, 100)), axis=0))

def sysidx(D):
    rmin = np.nanmin(D, axis=0)
    o = np.sort(D, axis=0)[1:] / 100.0
    return rmin * np.sqrt(np.nanprod(o, axis=0))

variants = {
    "A min-in-domain (current)": sysidx(np.vstack([np.nanmin(R[k],0) for k in F])),
    "B geomean-in-domain":       sysidx(np.vstack([gmean(R[k]) for k in F])),
    "C geomean of all factors":  gmean(np.vstack([R[k] for k in F])),
}
flat = d["slope_pct"].to_numpy(float) <= 4.0
print(f"flat cells (slope<=4%): {flat.sum():,} of {len(d):,} = {100*flat.mean():.1f}%\n")
for name, v in variants.items():
    s = lambda m: (f"S1 {100*(m>=75).mean():5.2f}%  S2 {100*((m>=50)&(m<75)).mean():5.2f}%"
                   f"  S3 {100*((m>=25)&(m<50)).mean():5.2f}%  N {100*(m<25).mean():5.2f}%")
    print(f"{name:28s} all  {s(v)}  p99={np.nanpercentile(v,99):.0f}")
    print(f"{'':28s} flat {s(v[flat])}\n")
```
